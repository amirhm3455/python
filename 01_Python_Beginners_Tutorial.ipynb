# جزوه ۱: آشنایی با پایتون

## ۱. مقدمه و چاپ اطلاعات
* پایتون یک زبان برنامه‌نویسی بسیار خوانا و شبیه به زبان انگلیسی است که برای شروع برنامه‌نویسی بهترین انتخاب محسوب می‌شود.
* اولین دستوری که یاد می‌گیریم `print` است. این دستور به ما کمک می‌کند تا متن‌ها، اعداد یا نتیجه محاسبات را روی صفحه نمایش دهیم.
* برای چاپ متن‌ها، حتما باید آن‌ها را داخل علامت کوتیشن (مثل "متن" یا 'متن') قرار دهیم.
* می‌توانیم چند مقدار را با کاما (`,`) جدا کنیم و در یک `print` چاپ کنیم؛ پایتون بین آن‌ها خودکار یک فاصله می‌گذارد.

</div>

```python
# استفاده از دستور پرینت برای نمایش متن
print("به دنیای برنامه‌نویسی پایتون خوش آمدید!")

# چاپ یک عدد بدون نیاز به کوتیشن
print(1405)

# چاپ نتیجه یک محاسبه
print(2 + 3)

# چاپ چند مقدار در یک خط (با کاما جدا می‌شوند)
print("حاصل جمع ۲ و ۳ برابر است با:", 2 + 3)
```

## ۲. شکستن خطوط طولانی کد
* در برنامه‌نویسی پایتون، اگر یک خط کد خیلی طولانی شود، خواندن آن برای ما سخت خواهد شد.
* پایتون به ما اجازه می‌دهد کدهای طولانی را بشکنیم و در چند خط بنویسیم تا ظاهر مرتب‌تری داشته باشند.
* بهترین و استانداردترین روش برای شکستن خطوط، قرار دادن کل آن عبارت داخل یک جفت پرانتز `()` است.
* روش دیگر استفاده از علامت بک‌اسلش `\` در انتهای خط است، اما چون ممکن است باعث خطا شود (مثلا اگر بعد از آن فاصله بگذاریم)، روش پرانتز توصیه می‌شود.

</div>

```python
# استفاده از پرانتز برای نوشتن یک محاسبه طولانی در چند خط
# این کار باعث می‌شود کد ما بسیار مرتب‌تر و خواناتر شود
total_sum = (
    10 + 20 +
    30 + 40 +
    50 + 60
)

# چاپ نتیجه نهایی محاسبه بالا
print(total_sum)
```

## ۳. توضیحات (کامنت‌گذاری) در کد
* کامنت‌ها یادداشت‌هایی هستند که ما در میان کدهای خود می‌نویسیم تا بعدا خودمان یا دیگران بفهمیم کد چه کاری انجام می‌دهد.
* رایانه هنگام اجرای برنامه، کامنت‌ها را کاملا نادیده می‌گیرد و آن‌ها هیچ تاثیری در خروجی برنامه ندارند.
* در زبان پایتون، برای نوشتن کامنت کافی است علامت هشتگ `#` را تایپ کنیم؛ هر چیزی که بعد از این علامت بیاید، توضیح محسوب می‌شود.
* برای کامنت چندخطی می‌توانیم از سه کوتیشن `"""` استفاده کنیم (بیشتر برای توضیح توابع کاربرد دارد).

</div>

```python
# این یک کامنت است و توسط پایتون اجرا نمی‌شود
name = "علی"  # می‌توانیم کامنت‌ها را در انتهای خط کد هم بنویسیم

# حالا نام ذخیره شده را نمایش می‌دهیم
print(name)

# کد زیر غیرفعال شده است، چون ابتدای آن # دارد:
# print("این خط اجرا نمی‌شود")
```

## ۴. متغیرها و قوانین نام‌گذاری
* متغیر مثل یک جعبه با برچسب است که اطلاعات را داخل آن نگه می‌داریم تا بعدا از آن استفاده کنیم.
* برای ساخت متغیر از علامت مساوی `=` استفاده می‌کنیم؛ نام متغیر سمت چپ و مقدار آن سمت راست قرار می‌گیرد.
* قوانین نام متغیر: فقط شامل حروف انگلیسی، عدد و زیرخط (`_`) باشد، با عدد شروع نشود و فاصله نداشته باشد.
* پایتون بین حروف بزرگ و کوچک فرق می‌گذارد؛ یعنی `age` و `Age` دو متغیر متفاوت هستند.
* بهتر است نام متغیرها معنی‌دار باشند (مثلا `student_age` بهتر از `a` است).

</div>

```python
# ساخت چند متغیر
student_name = "سارا"
student_age = 20

print(student_name, student_age)

# مقدار یک متغیر را می‌توانیم بعدا تغییر دهیم
student_age = 21
print("سن جدید:", student_age)

# حروف بزرگ و کوچک متفاوت هستند: این دو متغیر کاملا جدا از هم‌اند
score = 10
Score = 99
print(score, Score)
```

## ۵. فواصل و تو رفتگی‌ها
* در زبان‌های دیگر معمولا از آکولاد `{}` برای دسته‌بندی و بلوک‌بندی کدها استفاده می‌شود، اما پایتون از فاصله (Space) استفاده می‌کند.
* این فواصل که به آن‌ها تو رفتگی (Indentation) می‌گویند، اجباری هستند و ساختار برنامه ما را مشخص می‌کنند.
* به طور استاندارد، همیشه از ۴ عدد فاصله (Space) برای ایجاد این تو رفتگی‌ها در دستوراتی مثل شرط‌ها استفاده می‌کنیم.
* اگر تو رفتگی را رعایت نکنیم، پایتون خطای `IndentationError` می‌دهد.

</div>

```python
# تعریف متغیر سن برای بررسی
age = 20

# بررسی یک شرط ساده
# دقت کنید که بعد از علامت دو نقطه، خط بعدی باید تو رفتگی داشته باشد
if age >= 18:
    # این خط دارای ۴ فاصله تو رفتگی است تا مشخص شود متعلق به شرط بالا است
    print("شما اجازه ورود دارید.")
    print("این خط هم جزو همان بلوک شرطی است.")

# این خط تو رفتگی ندارد، پس جزو شرط نیست و همیشه اجرا می‌شود
print("پایان برنامه")
```

## ۶. مثال جامع
* در این بخش یک سناریوی ساده برای خوش‌آمدگویی به کاربر می‌نویسیم که تمام مفاهیم بالا را ترکیب می‌کند.
* در این کد ما از کامنت‌ها برای توضیح، از متغیرها برای ذخیره اطلاعات، از تو رفتگی برای شرط، از پرانتز برای شکستن خطوط و از دستور چاپ استفاده کرده‌ایم.

</div>

```python
# ۱. تعریف متغیرهای اولیه کاربر
user_name = "سارا"
user_score = 85

# ۲. بررسی نمره کاربر با استفاده از شرط و تو رفتگی
if user_score > 50:
    # ۳. تعریف یک پیام طولانی در چند خط با استفاده از پرانتز
    welcome_message = (
        "کاربر گرامی، "
        "شما در آزمون قبول شدید. "
        "به سطح بعدی خوش آمدید!"
    )

    # ۴. نمایش نام کاربر و پیام تبریک
    print(user_name)
    print(welcome_message)
```

## ۷. مثال‌های بیشتر
در این بخش چند مثال دیگر می‌بینیم تا مفاهیم این جزوه کاملا جا بیفتند. سعی کنید قبل از اجرا، خروجی هر مثال را حدس بزنید.

### مثال الف: معرفی یک دانشجو

</div>

```python
# متغیرهای مشخصات دانشجو
full_name = "رضا احمدی"
major = "هوش مصنوعی"
semester = 3

# چاپ چند مقدار در یک خط
print("نام:", full_name)
print("رشته:", major)
print("ترم:", semester)

# محاسبه ترمی که دانشجو در آن قرار می‌گیرد پس از دو ترم دیگر
print("دو ترم دیگر در ترم", semester + 2, "خواهد بود.")
```

```python
# دمای هوا به درجه سانتی‌گراد
temperature = 32

# بررسی دما و چاپ پیام مناسب
if temperature > 30:
    print("هوا گرم است.")
    print("آب زیاد بنوشید.")

print("گزارش آب‌وهوا تمام شد.")
```

### مثال ج: محاسبه قیمت با شکستن خط

</div>

```python
# قیمت سه کالا (به تومان)
price_1 = 120000
price_2 = 85000
price_3 = 43000

# جمع قیمت‌ها در چند خط با کمک پرانتز
final_price = (
    price_1 +
    price_2 +
    price_3
)

print("مبلغ قابل پرداخت:", final_price, "تومان")
```
