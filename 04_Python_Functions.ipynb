# جزوه ۴: توابع

## ۱. توابع چیستند و توابع آماده پایتون
* در برنامه‌نویسی، وقتی یک تکه کد را زیاد استفاده می‌کنیم، آن را داخل یک بسته‌بندی به نام «تابع» قرار می‌دهیم تا مجبور نباشیم کد را تکرار کنیم.
* توابع مثل کارخانه‌های کوچکی هستند که یک ورودی می‌گیرند، کاری روی آن انجام می‌دهند و یک خروجی می‌دهند.
* پایتون از قبل توابع آماده‌ای دارد که برنامه‌نویسان پایتون آن‌ها را ساخته‌اند؛ مثل دستور چاپ که قبلاً دیده‌اید، یا تابعی که طول یک متن را می‌شمارد.
* چند تابع آماده پرکاربرد: `len()` (تعداد اعضا)، `max()` و `min()` (بیشترین و کمترین)، `sum()` (جمع)، `sorted()` (مرتب‌شده)، `round()` (گرد کردن)، `abs()` (قدر مطلق) و `type()` (نوع داده).

</div>

```python
# استفاده از توابع آماده پایتون
text = "سلام به پایتون"

# تابع len برای شمردن تعداد کاراکترها (حروف و فاصله‌ها) استفاده می‌شود
text_length = len(text)

# تابع print برای نمایش اطلاعات استفاده می‌شود
print("طول متن ما برابر است با:")
print(text_length)

print("-----------------")

# چند تابع آماده دیگر
numbers = [8, -3, 15, 4]
print("بیشترین:", max(numbers))
print("کمترین:", min(numbers))
print("جمع:", sum(numbers))
print("مرتب‌شده:", sorted(numbers))
print("قدر مطلق:", abs(-7))
print("گرد شده:", round(3.14159, 2))
```

## ۲. ساخت و استفاده از توابع (تعریف و فراخوانی)
* برای ساختن یک تابع اختصاصی، از کلمه کلیدی `def` (مخفف define به معنی تعریف کردن) استفاده می‌کنیم.
* بعد از این کلمه، نام تابع را می‌نویسیم، یک پرانتز باز و بسته می‌گذاریم و در انتها علامت دو نقطه `:` قرار می‌دهیم.
* کدهایی که داخل تابع هستند باید تو رفتگی (فاصله از ابتدای خط) داشته باشند.
* توابع تا زمانی که آن‌ها را صدا نزنیم (فراخوانی نکنیم)، هیچ کاری انجام نمی‌دهند. برای صدا زدن، فقط کافی است نام آن‌ها را همراه با پرانتز بنویسیم.

</div>

```python
# تعریف یک تابع ساده برای احوال‌پرسی
def say_hello():
    # این کدها فقط زمانی اجرا می‌شوند که تابع صدا زده شود
    print("سلام!")
    print("به برنامه ما خوش آمدید.")

# فراخوانی (صدا زدن) تابع
# با این کار، کدهای داخل تابع به ترتیب اجرا می‌شوند
say_hello()

# هر چند بار بخواهیم می‌توانیم تابع را دوباره صدا بزنیم
say_hello()
```

## ۳. ارسال اطلاعات به تابع (آرگومان‌ها)
* توابع می‌توانند هوشمندتر باشند و بر اساس اطلاعاتی که ما به آن‌ها می‌دهیم، کارهای متفاوتی انجام دهند.
* به متغیرهایی که در پرانتزِ تعریف تابع می‌نویسیم تا اطلاعات را دریافت کنند، «پارامتر» می‌گویند.
* به مقادیری که هنگام صدا زدن تابع به آن می‌دهیم تا جایگزین پارامترها شوند، «آرگومان» می‌گویند.
* با این کار، یک تابع می‌تواند برای نام‌ها یا اعداد مختلف، بی‌نهایت بار استفاده شود.

</div>

```python
# تعریف تابعی که یک ورودی (نام) دریافت می‌کند
def greet_user(name):
    # از ورودی دریافت شده در متن چاپ استفاده می‌کنیم
    print("کاربر گرامی", name, "خوش آمدید.")

# صدا زدن تابع و ارسال مقادیر مختلف (آرگومان‌ها)
greet_user("علی")
greet_user("سارا")
```

## ۴. دریافت نتیجه از تابع (دستور return)
* دستور چاپ فقط اطلاعات را روی صفحه نمایش می‌دهد، اما گاهی ما می‌خواهیم نتیجه محاسبه تابع را نگه داریم تا در ادامه برنامه از آن استفاده کنیم.
* دستور `return` (برگرداندن) به تابع می‌گوید: «نتیجه نهایی را به من برگردان تا بتوانم آن را در یک متغیر ذخیره کنم».
* وقتی تابع به کلمه `return` می‌رسد، کارش تمام می‌شود و بلافاصله از تابع خارج می‌شود.
* یک تابع می‌تواند چند مقدار را با هم برگرداند؛ پایتون آن‌ها را به صورت تاپل برمی‌گرداند و ما می‌توانیم در چند متغیر باز کنیم.
* تابعی که `return` نداشته باشد، مقدار `None` (یعنی هیچ) برمی‌گرداند.

</div>

```python
# تعریف تابعی برای جمع کردن دو عدد
def add_numbers(num1, num2):
    total = num1 + num2
    # برگرداندن نتیجه به جای چاپ کردن آن
    return total

# ذخیره نتیجه تابع در یک متغیر
result = add_numbers(10, 15)

# حالا می‌توانیم از این نتیجه در جاهای دیگر برنامه استفاده کنیم
print("حاصل جمع برابر است با:", result)
print("اگر نتیجه را دو برابر کنیم می‌شود:", result * 2)

print("-----------------")

# برگرداندن چند مقدار با هم
def min_and_max(numbers):
    return min(numbers), max(numbers)

smallest, biggest = min_and_max([4, 9, 1, 7])
print("کمترین:", smallest, "| بیشترین:", biggest)
```

## ۵. مقادیر پیش‌فرض
* گاهی می‌خواهیم اگر کاربر به تابع ورودی نداد، برنامه خطا ندهد و از یک مقدار پیش‌فرض استفاده کند. برای این کار در پرانتزِ تعریف تابع، مقدار پیش‌فرض را با علامت مساوی مشخص می‌کنیم.
* پارامترهای دارای مقدار پیش‌فرض باید **بعد از** پارامترهای معمولی نوشته شوند.

</div>

```python
def order_coffee(size="کوچک"):
    print("شما یک قهوه با اندازه", size, "سفارش دادید.")

# اگر ورودی بدهیم، همان ورودی استفاده می‌شود
order_coffee("بزرگ")
# اگر ورودی ندهیم، از مقدار پیش‌فرض استفاده می‌شود
order_coffee()

print("-----------------")

# ترتیب درست: پارامتر معمولی اول، پارامتر پیش‌فرض بعد
def power(base, exponent=2):
    return base ** exponent

print(power(5))      # توان پیش‌فرض ۲ است
print(power(2, 3))   # توان ۳
```

## ۶. ورودی‌های نامحدود (`*args`)
* اگر ندانیم کاربر دقیقاً چند ورودی می‌خواهد به تابع بدهد، قبل از نام پارامتر یک ستاره `*` می‌گذاریم.
* با این کار تمام ورودی‌ها در یک **تاپل** (همان ساختار داده‌ای که در جزوه ۲ دیدیم) جمع‌آوری می‌شوند و می‌توانیم روی آن‌ها حلقه بزنیم.

</div>

```python
# دریافت تعداد نامحدودی ورودی
def show_all_friends(*names):
    # کلمه names اکنون یک تاپل شامل تمام ورودی‌ها است و می‌توانیم روی آن حلقه بزنیم
    print("نوع names:", type(names))
    for friend in names:
        print("دوست شما:", friend)

show_all_friends("رضا", "مریم", "امید")

print("-----------------")

# تابع جمع چند عدد
def sum_all(*numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(sum_all(1, 2, 3))
print(sum_all(10, 20, 30, 40, 50))
```

## ۷. آرگومان‌های کلیدی (Keyword Arguments)
* هنگام صدا زدن تابع می‌توانیم نام پارامتر را هم بنویسیم (`name=مقدار`). به این کار آرگومان کلیدی می‌گویند و باعث می‌شود کد خواناتر شود و ترتیب ورودی‌ها مهم نباشد.
* اگر بعد از `*args` پارامتری با مقدار پیش‌فرض بنویسیم، آن پارامتر را **فقط** با نام می‌توان مقداردهی کرد. این روش درست برای ترکیب ورودی نامحدود و مقدار پیش‌فرض است (در مثال جامع از آن استفاده می‌کنیم).

</div>

```python
def introduce(name, age, city="تهران"):
    print(name, "،", age, "ساله، ساکن", city)

# ترتیب ورودی‌ها وقتی نام می‌نویسیم مهم نیست
introduce(age=21, name="سارا")
introduce("علی", 25, city="شیراز")

print("-----------------")

# ترکیب *args و آرگومان کلیدی
def join_words(*words, separator=" "):
    return separator.join(words)

print(join_words("من", "پایتون", "می‌خوانم"))
print(join_words("الف", "ب", "ج", separator="-"))
```

## ۸. مستندسازی توابع (Docstrings)
* وقتی کدهای ما طولانی می‌شوند، باید برای توابعی که می‌نویسیم یک راهنمای کوچک بسازیم تا مشخص شود این تابع چه وظیفه‌ای دارد.
* این کار در پایتون با قرار دادن سه علامت کوتیشن `"""` در اولین خطِ داخل تابع انجام می‌شود.
* این توضیحات توسط پایتون اجرا نمی‌شوند و فقط برای راهنمایی برنامه‌نویسان هستند. با دستور `help(نام_تابع)` می‌توان آن را دید.

</div>

```python
# تعریف تابع همراه با توضیحات راهنما (Docstring)
def calculate_area(length, width):
    """
    این تابع طول و عرض یک مستطیل را دریافت می‌کند
    و مساحت آن را محاسبه کرده و برمی‌گرداند.
    """
    area = length * width
    return area

# استفاده از تابع
print("مساحت مستطیل:", calculate_area(5, 4))

# نمایش راهنمای تابع
help(calculate_area)
```

## ۹. متغیرهای محلی و سراسری (Scope)
* متغیری که **داخل** تابع ساخته می‌شود «محلی» است و فقط همان‌جا وجود دارد؛ بعد از پایان تابع از بین می‌رود.
* متغیری که **بیرون** از تابع ساخته می‌شود «سراسری» است و تابع می‌تواند آن را بخواند.
* بهترین روش این است که اطلاعات را با پارامتر به تابع بدهیم و نتیجه را با `return` بگیریم، نه اینکه به متغیرهای بیرونی وابسته باشیم.

</div>

```python
tax_rate = 0.09      # متغیر سراسری

def price_with_tax(price):
    tax = price * tax_rate     # tax یک متغیر محلی است
    return price + tax

print(price_with_tax(1000))

# خط زیر خطا می‌دهد، چون tax فقط داخل تابع وجود داشت:
# print(tax)
```

## ۱۰. توابع یک‌خطی (Lambda)
* در پایتون راهی وجود دارد که توابع بسیار کوچک و ساده ریاضی را به جای استفاده از دستور `def`، فقط در یک خط تعریف کنیم.
* به این توابع «لامبدا» می‌گویند. این روش با کلمه `lambda` شروع می‌شود، سپس ورودی‌ها و بعد از علامت دو نقطه، عملیات مورد نظر قرار می‌گیرد.
* لامبدا ذاتاً بدون نام است؛ ما در مثال اول برای اینکه چند بار استفاده کنیم آن را در یک متغیر ذخیره کرده‌ایم. کاربرد اصلی آن همین است که بدون نام و مستقیم داخل توابعی مثل `sorted` به کار برود.
* این روش فقط برای کارهای بسیار کوتاه کاربرد دارد و در ابتدای مسیر یادگیری، آشنایی با شکل ظاهری آن کافی است.

</div>

```python
# تعریف یک تابع لامبدا برای دو برابر کردن یک عدد
# کلمه x نقش ورودی را دارد و x * 2 کاری است که روی آن انجام می‌شود
double_number = lambda x: x * 2

# استفاده از تابع لامبدا
print("دو برابر عدد ۵ می‌شود:", double_number(5))

# لامبدا با دو ورودی
multiply = lambda a, b: a * b
print(multiply(3, 4))

# کاربرد اصلی: مشخص کردن معیار مرتب‌سازی
students = [("علی", 17), ("سارا", 20), ("رضا", 15)]
# بر اساس عضو دوم هر تاپل (یعنی نمره) مرتب می‌کنیم
by_score = sorted(students, key=lambda s: s[1])
print(by_score)
```

## ۱۱. مثال جامع
* در این سناریو، یک سیستم ساده برای محاسبه قیمت نهایی محصولات یک سبد خرید فروشگاهی می‌نویسیم.
* در این تابع ما از تمام مفاهیم این جزوه استفاده کرده‌ایم: تعریف تابع (`def`)، مستندسازی، دریافت ورودی نامحدود برای قیمت‌ها (`*prices`)، دریافت مقدار پیش‌فرض برای تخفیف به صورت آرگومان کلیدی (`discount=0`) و برگرداندن خروجی نهایی (`return`).
* دقت کنید که `discount` بعد از `*prices` آمده است؛ به همین دلیل مقدار پیش‌فرض آن واقعاً کار می‌کند و اگر تخفیف ندهیم، برابر ۰ می‌ماند.

</div>

```python
# تعریف تابع جامع برای محاسبه قیمت کل فروشگاه
def calculate_total_price(*prices, discount=0):
    """
    این تابع قیمت کالاها را (به تعداد دلخواه) دریافت می‌کند،
    درصد تخفیف (در صورت وجود) را اعمال می‌کند،
    و مبلغ نهایی قابل پرداخت را برمی‌گرداند.
    """
    # ابتدا مجموع تمام قیمت‌های وارد شده را محاسبه می‌کنیم
    total = 0
    for price in prices:
        total = total + price

    # محاسبه مقدار تخفیف (اگر تخفیف داده شده باشد)
    discount_amount = (discount / 100) * total

    # کسر تخفیف از مبلغ کل
    final_price = total - discount_amount

    # برگرداندن قیمت نهایی
    return final_price

# محاسبه قیمت چند کالا (بدون تخفیف؛ مقدار پیش‌فرض ۰ استفاده می‌شود)
bill_1 = calculate_total_price(100, 200, 50)
print("فاکتور اول (بدون تخفیف):", bill_1)

# محاسبه قیمت چند کالا (با ۲۰ درصد تخفیف)
# تخفیف را با نام مشخص می‌کنیم و بقیه اعداد قیمت کالاها هستند
bill_2 = calculate_total_price(100, 200, 50, discount=20)
print("فاکتور دوم (با ۲۰ درصد تخفیف):", bill_2)
```

## ۱۲. مثال‌های بیشتر
### مثال الف: تابع زوج یا فرد

</div>

```python
def is_even(number):
    """اگر عدد زوج باشد True و در غیر این صورت False برمی‌گرداند."""
    return number % 2 == 0

print(is_even(10))
print(is_even(7))

# استفاده از تابع داخل یک حلقه
for n in range(1, 6):
    if is_even(n):
        print(n, "زوج است")
    else:
        print(n, "فرد است")
```

### مثال ب: میانگین یک لیست

</div>

```python
def average(numbers):
    """میانگین اعداد یک لیست را برمی‌گرداند. برای لیست خالی صفر برمی‌گرداند."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average([17, 14, 19]))
print(average([]))
```

### مثال ج: تبدیل دما

</div>

```python
def celsius_to_fahrenheit(c):
    """درجه سانتی‌گراد را به فارنهایت تبدیل می‌کند."""
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f):
    """درجه فارنهایت را به سانتی‌گراد تبدیل می‌کند."""
    return (f - 32) * 5 / 9

print("۱۰۰ سانتی‌گراد برابر است با", celsius_to_fahrenheit(100), "فارنهایت")
print("۹۸.۶ فارنهایت برابر است با", round(fahrenheit_to_celsius(98.6), 1), "سانتی‌گراد")
```

### مثال د: تعیین وضعیت نمره با تابع و شرط

</div>

```python
def get_grade(score):
    """بر اساس نمره، وضعیت را به صورت متن برمی‌گرداند."""
    if score >= 90:
        return "عالی"
    elif score >= 70:
        return "خوب"
    elif score >= 50:
        return "قبول"
    else:
        return "مردود"

scores = [95, 72, 55, 30]
for s in scores:
    print("نمره", s, "->", get_grade(s))
```

### مثال ه: شمارش حروف صدادار در یک متن انگلیسی

</div>

```python
def count_vowels(text):
    """تعداد حروف صدادار انگلیسی یک متن را می‌شمارد."""
    vowels = "aeiou"
    count = 0
    for letter in text.lower():
        if letter in vowels:
            count += 1
    return count

print(count_vowels("Artificial Intelligence"))
```
