## ۱. توابع چیستند و توابع آماده پایتون
* در برنامه‌نویسی، وقتی یک تکه کد را زیاد استفاده می‌کنیم، آن را داخل یک بسته‌بندی به نام «تابع» قرار می‌دهیم تا مجبور نباشیم کد را تکرار کنیم.
* توابع مثل کارخانه‌های کوچکی هستند که یک ورودی می‌گیرند، کاری روی آن انجام می‌دهند و یک خروجی می‌دهند.
* پایتون از قبل توابع آماده‌ای دارد که برنامه‌نویسان پایتون آن‌ها را ساخته‌اند؛ مثل دستور چاپ که قبلاً دیده‌اید، یا تابعی که طول یک متن را می‌شمارد.

</div>

```python
# استفاده از توابع آماده پایتون
text = "سلام به پایتون"

# تابع len برای شمردن تعداد کاراکترها (حروف و فاصله‌ها) استفاده می‌شود
text_length = len(text)

# تابع print برای نمایش اطلاعات استفاده می‌شود
print("طول متن ما برابر است با:")
print(text_length)
```

## ۲. ساخت و استفاده از توابع (تعریف و فراخوانی)
* برای ساختن یک تابع اختصاصی، از کلمه کلیدی `def` (مخفف define به معنی تعریف کردن) استفاده می‌کنیم.
* بعد از این کلمه، نام تابع را می‌نویسیم، یک پرانتز باز و بسته می‌گذاریم و در انتها علامت دو نقطه `:` قرار می‌دهیم.
* کدهایی که داخل تابع هستند باید تو رفتگی (فاصله از ابتدای خط) داشته باشند.
* توابع تا زمانی که آن‌ها را صدا نزنیم (فراخوانی نکنیم)، هیچ کاری انجام نمی‌دهند. برای صدا زدن، فقط کافی است نام آن‌ها را همراه با پرانتز بنویسیم.

</div>

```python
# تعریف یک تابع ساده برای احوال‌پرسی
def say_hello():
    # این کدها فقط زمانی اجرا می‌شوند که تابع صدا زده شود
    print("سلام!")
    print("به برنامه ما خوش آمدید.")

# فراخوانی (صدا زدن) تابع
# با این کار، کدهای داخل تابع به ترتیب اجرا می‌شوند
say_hello()
```

## ۳. ارسال اطلاعات به تابع (آرگومان‌ها)
* توابع می‌توانند هوشمندتر باشند و بر اساس اطلاعاتی که ما به آن‌ها می‌دهیم، کارهای متفاوتی انجام دهند.
* به متغیرهایی که در پرانتزِ تعریف تابع می‌نویسیم تا اطلاعات را دریافت کنند، «پارامتر» می‌گویند.
* به مقادیری که هنگام صدا زدن تابع به آن می‌دهیم تا جایگزین پارامترها شوند، «آرگومان» می‌گویند.
* با این کار، یک تابع می‌تواند برای نام‌ها یا اعداد مختلف، بی‌نهایت بار استفاده شود.

</div>

```python
# تعریف تابعی که یک ورودی (نام) دریافت می‌کند
def greet_user(name):
    # از ورودی دریافت شده در متن چاپ استفاده می‌کنیم
    print("کاربر گرامی", name, "خوش آمدید.")

# صدا زدن تابع و ارسال مقادیر مختلف (آرگومان‌ها)
greet_user("علی")
greet_user("سارا")
```

## ۴. دریافت نتیجه از تابع (دستور return)
* دستور چاپ فقط اطلاعات را روی صفحه نمایش می‌دهد، اما گاهی ما می‌خواهیم نتیجه محاسبه تابع را نگه داریم تا در ادامه برنامه از آن استفاده کنیم.
* دستور `return` (برگرداندن) به تابع می‌گوید: «نتیجه نهایی را به من برگردان تا بتوانم آن را در یک متغیر ذخیره کنم».
* وقتی تابع به کلمه `return` می‌رسد، کارش تمام می‌شود و بلافاصله از تابع خارج می‌شود.

</div>

```python
# تعریف تابعی برای جمع کردن دو عدد
def add_numbers(num1, num2):
    total = num1 + num2
    # برگرداندن نتیجه به جای چاپ کردن آن
    return total

# ذخیره نتیجه تابع در یک متغیر
result = add_numbers(10, 15)

# حالا می‌توانیم از این نتیجه در جاهای دیگر برنامه استفاده کنیم
print("حاصل جمع برابر است با:", result)
print("اگر نتیجه را دو برابر کنیم می‌شود:", result * 2)
```

## ۵. مقادیر پیش‌فرض و ورودی‌های متغیر
* گاهی می‌خواهیم اگر کاربر به تابع ورودی نداد، برنامه خطا ندهد و از یک مقدار پیش‌فرض استفاده کند. برای این کار در پرانتزِ تعریف تابع، مقدار پیش‌فرض را با علامت مساوی مشخص می‌کنیم.
* همچنین اگر ندانیم کاربر دقیقاً چند ورودی می‌خواهد به تابع بدهد، قبل از نام پارامتر یک ستاره `*` می‌گذاریم. با این کار تمام ورودی‌ها در یک مجموعه جمع‌آوری می‌شوند.

</div>

```python
# ۱. مثال برای مقدار پیش‌فرض
def order_coffee(size="کوچک"):
    print("شما یک قهوه با اندازه", size, "سفارش دادید.")

# اگر ورودی بدهیم، همان ورودی استفاده می‌شود
order_coffee("بزرگ")
# اگر ورودی ندهیم، از مقدار پیش‌فرض استفاده می‌شود
order_coffee()

print("-----------------")

# ۲. مثال برای دریافت تعداد نامحدودی ورودی
def show_all_friends(*names):
    # کلمه names اکنون شامل تمام ورودی‌ها است و می‌توانیم روی آن حلقه بزنیم
    for friend in names:
        print("دوست شما:", friend)

show_all_friends("رضا", "مریم", "امید")
```

## ۶. مستندسازی توابع (Docstrings)
* وقتی کدهای ما طولانی می‌شوند، باید برای توابعی که می‌نویسیم یک راهنمای کوچک بسازیم تا مشخص شود این تابع چه وظیفه‌ای دارد.
* این کار در پایتون با قرار دادن سه علامت کوتیشن `"""` در اولین خطِ داخل تابع انجام می‌شود.
* این توضیحات توسط پایتون اجرا نمی‌شوند و فقط برای راهنمایی برنامه‌نویسان هستند.

</div>

```python
# تعریف تابع همراه با توضیحات راهنما (Docstring)
def calculate_area(length, width):
    """
    این تابع طول و عرض یک مستطیل را دریافت می‌کند
    و مساحت آن را محاسبه کرده و برمی‌گرداند.
    """
    area = length * width
    return area

# استفاده از تابع
print("مساحت مستطیل:", calculate_area(5, 4))
```

## ۷. توابع یک‌خطی (Lambda)
* در پایتون راهی وجود دارد که توابع بسیار کوچک و ساده ریاضی را به جای استفاده از دستور `def`، فقط در یک خط تعریف کنیم.
* به این توابع بدون نام، «لامبدا» می‌گویند. این روش با کلمه `lambda` شروع می‌شود، سپس ورودی‌ها و بعد از علامت دو نقطه، عملیات مورد نظر قرار می‌گیرد.
* این روش فقط برای کارهای بسیار کوتاه کاربرد دارد و در ابتدای مسیر یادگیری، آشنایی با شکل ظاهری آن کافی است.

</div>

```python
# تعریف یک تابع لامبدا برای دو برابر کردن یک عدد
# کلمه x نقش ورودی را دارد و x * 2 کاری است که روی آن انجام می‌شود
double_number = lambda x: x * 2

# استفاده از تابع لامبدا
print("دو برابر عدد ۵ می‌شود:", double_number(5))
```

## ۸. مثال جامع
* در این سناریو، یک سیستم ساده برای محاسبه قیمت نهایی محصولات یک سبد خرید فروشگاهی می‌نویسیم.
* در این تابع ما از تمام مفاهیم این جزوه استفاده کرده‌ایم: تعریف تابع (`def`)، مستندسازی، دریافت ورودی نامحدود برای قیمت‌ها، دریافت مقدار پیش‌فرض برای تخفیف و برگرداندن خروجی نهایی (`return`).

</div>

```python
# تعریف تابع جامع برای محاسبه قیمت کل فروشگاه
def calculate_total_price(discount=0, *prices):
    """
    این تابع مجموع قیمت کالاها را می‌گیرد،
    درصد تخفیف (در صورت وجود) را اعمال می‌کند،
    و مبلغ نهایی قابل پرداخت را برمی‌گرداند.
    """
    # ابتدا مجموع تمام قیمت‌های وارد شده را محاسبه می‌کنیم
    total = 0
    for price in prices:
        total = total + price
        
    # محاسبه مقدار تخفیف (اگر تخفیف داده شده باشد)
    discount_amount = (discount / 100) * total
    
    # کسر تخفیف از مبلغ کل
    final_price = total - discount_amount
    
    # برگرداندن قیمت نهایی
    return final_price

# محاسبه قیمت چند کالا (بدون تخفیف)
bill_1 = calculate_total_price(0, 100, 200, 50)
print("فاکتور اول (بدون تخفیف):", bill_1)

# محاسبه قیمت چند کالا (با ۲۰ درصد تخفیف)
# عدد اول نشان‌دهنده درصد تخفیف و اعداد بعدی قیمت کالاها هستند
bill_2 = calculate_total_price(20, 100, 200, 50)
print("فاکتور دوم (با ۲۰ درصد تخفیف):", bill_2)
```
