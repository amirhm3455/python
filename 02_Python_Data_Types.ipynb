## ۱. انواع داده‌ها و متغیرها
در پایتون متغیر مثل برچسبی است که روی یک جعبه می‌زنیم تا نوع اطلاعات درون آن را مشخص کنیم:
- **اعداد صحیح (int):** اعداد بدون اعشار مانند `25`, `0`, `-5`
- **اعداد اعشاری (float):** اعداد دارای اعشار مانند `1.75`, `19.99`
- **رشته‌ها (str):** متن‌ها که داخل کوتیشن قرار می‌گیرند مانند `"Python"`
- **بولی (bool):** مقادیر منطقی که فقط دو حالت دارند: درست (`True`) یا نادرست (`False`)
- با تابع `type()` می‌توانیم نوع هر مقدار را ببینیم.

</div>

```python
# مثال ۱: تعریف متغیرها و بررسی نوع آن‌ها
course_name = "آموزش پایتون"   # str
students_count = 15            # int
average_score = 18.5           # float
is_active = True               # bool

# چاپ مقدار و نوع متغیر
print("نام دوره:", course_name, "| نوع:", type(course_name))
print("تعداد:", students_count, "| نوع:", type(students_count))
print("میانگین:", average_score, "| نوع:", type(average_score))
print("وضعیت:", is_active, "| نوع:", type(is_active))
```

## ۲. عملگرهای ریاضی
عملگر (Operator) علامتی است که یک عمل روی مقادیر انجام می‌دهد. عملگرهای ریاضی پایتون:

| عملگر | معنی | مثال | نتیجه |
| :---: | :--- | :---: | :---: |
| `+` | جمع | `7 + 2` | `9` |
| `-` | تفریق | `7 - 2` | `5` |
| `*` | ضرب | `7 * 2` | `14` |
| `/` | تقسیم (نتیجه همیشه اعشاری) | `7 / 2` | `3.5` |
| `//` | تقسیم صحیح (فقط بخش صحیح خارج‌قسمت) | `7 // 2` | `3` |
| `%` | باقیمانده تقسیم | `7 % 2` | `1` |
| `**` | توان | `7 ** 2` | `49` |

* ترتیب انجام محاسبات مثل ریاضی است: اول پرانتز، بعد توان، بعد ضرب و تقسیم، و در آخر جمع و تفریق.
* عملگر `%` کاربرد زیادی دارد؛ مثلا اگر `x % 2` صفر شود، عدد `x` زوج است.

</div>

```python
a = 7
b = 2

print("جمع:", a + b)
print("تفریق:", a - b)
print("ضرب:", a * b)
print("تقسیم:", a / b)
print("تقسیم صحیح:", a // b)
print("باقیمانده:", a % b)
print("توان:", a ** b)

# اولویت عملگرها: اول ضرب انجام می‌شود
print(2 + 3 * 4)      # برابر 14
# با پرانتز اولویت را تغییر می‌دهیم
print((2 + 3) * 4)    # برابر 20

# کاربرد باقیمانده: زوج یا فرد بودن
print("آیا ۱۰ زوج است؟", 10 % 2 == 0)
```

## ۳. عملگرهای انتساب
* علامت `=` مقدار را در متغیر قرار می‌دهد (انتساب).
* برای تغییر مقدار یک متغیر بر اساس مقدار قبلی‌اش، شکل کوتاه‌شده عملگرها را داریم:

| شکل کوتاه | معادل |
| :---: | :---: |
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |

</div>

```python
x = 10

x += 5     # x می‌شود 15
print(x)

x -= 3     # x می‌شود 12
print(x)

x *= 2     # x می‌شود 24
print(x)

x /= 4     # x می‌شود 6.0
print(x)
```

## ۴. رشته‌ها (String)
* رشته دنباله‌ای از کاراکترهاست و مثل لیست ایندکس دارد (شمارش از ۰ شروع می‌شود).
* با `+` دو رشته را به هم می‌چسبانیم و با `*` یک رشته را تکرار می‌کنیم.
* با `[شروع:پایان]` می‌توانیم بخشی از رشته را جدا کنیم (برش یا Slicing)؛ اندیس پایان شامل نمی‌شود.
* چند تابع مفید: `len()` برای طول، `.upper()` و `.lower()` برای تغییر حروف انگلیسی به بزرگ و کوچک، `.replace()` برای جایگزینی.
* عملگر `in` بررسی می‌کند که یک متن داخل متن دیگر هست یا نه.

</div>

```python
word = "Python"

print("طول:", len(word))
print("حرف اول:", word[0])
print("حرف آخر:", word[-1])      # ایندکس منفی از انتها شمرده می‌شود
print("سه حرف اول:", word[0:3])
print("بزرگ:", word.upper())
print("کوچک:", word.lower())

# چسباندن و تکرار رشته‌ها
print("Hello " + "World")
print("-" * 20)

# جایگزینی و جستجو
sentence = "من پایتون را دوست دارم"
print(sentence.replace("پایتون", "برنامه‌نویسی"))
print("آیا کلمه پایتون هست؟", "پایتون" in sentence)
```

## ۵. ساختارهای داده: لیست‌ها (List)

- **ساختار:** با براکت `[...]` ساخته می‌شود.
- **ویژگی‌ها:** مرتب است (ایندکس دارد)، مقادیر تکراری قبول می‌کند و **قابل تغییر (Mutable)** است.
- **کاربرد:** برای ذخیره مجموعه‌ای از داده‌ها که ممکن است در طول برنامه تغییر کنند.
- متدهای پرکاربرد: `append()` (اضافه به انتها)، `insert()` (اضافه در جای مشخص)، `remove()` (حذف یک مقدار)، `sort()` (مرتب‌سازی) و `len()` (تعداد اعضا).

</div>

```python
# مثال ۲: کار با لیست‌ها
fruits = ["سیب", "موز", "پرتقال"]

# دسترسی به آیتم‌ها (شمارش از ۰ شروع می‌شود)
print("میوه اول:", fruits[0])

# اضافه کردن عضو جدید به انتهای لیست
fruits.append("کیوی")

# تغییر دادن مقدار یک عضو
fruits[1] = "گلابی"

print("لیست نهایی:", fruits)

# چند عملیات دیگر
fruits.remove("سیب")          # حذف یک مقدار
print("بعد از حذف:", fruits)
print("تعداد اعضا:", len(fruits))

numbers = [5, 2, 9, 1]
numbers.sort()                # مرتب‌سازی از کوچک به بزرگ
print("مرتب‌شده:", numbers)
print("دو عضو اول:", numbers[0:2])   # برش لیست مثل رشته‌ها
```

## ۶. ساختارهای داده: تاپل‌ها (Tuple)

- **ساختار:** با پرانتز `(...)` ساخته می‌شود.
- **ویژگی‌ها:** شبیه لیست است، اما **غیرقابل تغییر (Immutable)** است.
- **کاربرد:** برای داده‌هایی که ثابت هستند و نباید تصادفاً تغییر کنند (مانند ابعاد تصویر یا مختصات جغرافیایی).
- می‌توانیم اعضای تاپل را در چند متغیر جداگانه باز کنیم (Unpacking).

</div>

```python
# مثال ۳: کار با تاپل‌ها
dimensions = (1920, 1080)

print("عرض تصویر:", dimensions[0])
print("ارتفاع تصویر:", dimensions[1])

# باز کردن تاپل در چند متغیر
width, height = dimensions
print("عرض:", width, "| ارتفاع:", height)

# کدهای زیر در صورت اجرا خطا می‌دهند چون تاپل قابل تغییر نیست:
# dimensions[0] = 1280
# dimensions.append(720)
```

## ۷. ساختارهای داده: مجموعه‌ها (Set)

- **ساختار:** با آکولاد `{...}` ساخته می‌شود.
- **ویژگی‌ها:** ترتیب مشخصی ندارد و **عضو تکراری را نمی‌پذیرد**.
- **کاربرد:** فیلتر کردن داده‌های تکراری و عملیات ریاضی مثل اشتراک و اجتماع.
- **نکته مهم:** آکولاد خالی `{}` یک **دیکشنری** خالی می‌سازد، نه مجموعه! برای ساخت مجموعه خالی باید `set()` بنویسیم.
- عملیات مجموعه‌ها: اجتماع با `|`، اشتراک با `&` و تفاضل با `-`.

</div>

```python
# مثال ۴: حذف داده‌های تکراری با استفاده از مجموعه
numbers = [1, 2, 2, 3, 4, 4, 4, 5]

# تبدیل لیست به مجموعه برای حذف خودکار تکراری‌ها
unique_numbers = set(numbers)

print("لیست اولیه (با تکرار):", numbers)
print("مجموعه (بدون تکرار):", unique_numbers)

# عملیات ریاضی روی مجموعه‌ها
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("اجتماع:", set_a | set_b)
print("اشتراک:", set_a & set_b)
print("تفاضل (فقط در A):", set_a - set_b)

# نکته: آکولاد خالی دیکشنری است
empty_dict = {}
empty_set = set()
print(type(empty_dict), type(empty_set))
```

## ۸. ساختارهای داده: دیکشنری‌ها (Dictionary)

- **ساختار:** با آکولاد و به صورت جفت‌های `key: value` (کلید: مقدار) ساخته می‌شود.
- **ویژگی‌ها:** به جای ایندکس عددی، با استفاده از نام "کلید" به مقادیر دسترسی پیدا می‌کنیم. کلیدها تکراری نیستند. ترتیب اضافه شدن اعضا حفظ می‌شود (از پایتون ۳.۷ به بعد)، اما با شماره ایندکس به اعضا دسترسی نداریم.
- **کاربرد:** ذخیره اطلاعات مرتبط با یک موجودیت (مثل مشخصات یک فرد یا محصول).
- متدهای پرکاربرد: `.keys()` (همه کلیدها)، `.values()` (همه مقدارها)، `.items()` (جفت‌ها) و `.get()` (خواندن امن مقدار).

</div>

```python
# مثال ۵: ذخیره مشخصات در دیکشنری
student = {
    "name": "علی",
    "age": 22,
    "major": "نرم‌افزار"
}

# خواندن مقدار با استفاده از کلید
print("نام دانشجو:", student["name"])

# اضافه کردن فیلد اطلاعاتی جدید
student["university"] = "تهران"

# تغییر مقدار قبلی
student["age"] = 23

print("اطلاعات کامل دانشجو:", student)

# متدهای پرکاربرد
print("کلیدها:", list(student.keys()))
print("مقدارها:", list(student.values()))

# اگر کلید وجود نداشته باشد، student["phone"] خطا می‌دهد؛ ولی get مقدار پیش‌فرض برمی‌گرداند
print("تلفن:", student.get("phone", "ثبت نشده"))

# پیمایش جفت‌های کلید و مقدار
for key, value in student.items():
    print(key, "=", value)
```

## ۱۰. مثال جامع
در این مثال ترکیبی، لیستی از دیکشنری‌ها داریم که نمایانگر اعضای یک تیم است. می‌خواهیم اسامی تکراری را با کمک مجموعه (Set) استخراج کنیم.

</div>

```python
# مثال ۶: ترکیب لیست، دیکشنری و مجموعه
team_members = [
    {"name": "رضا", "role": "توسعه‌دهنده"},
    {"name": "سارا", "role": "طراح"},
    {"name": "رضا", "role": "توسعه‌دهنده"} # این داده تکراری ثبت شده است
]

# استخراج اسامی از داخل دیکشنری‌ها و قرار دادن در یک لیست
names = []
for member in team_members:
    names.append(member["name"])

# تبدیل به مجموعه برای حذف موارد تکراری
unique_names = set(names)

print("لیست کل اسامی:", names)
print("اسامی یکتا بدون تکرار:", unique_names)
```

## ۱۱. مثال‌های بیشتر
### مثال الف: میانگین نمرات با لیست و عملگرها

</div>

```python
# مثال ۷: محاسبه میانگین یک لیست از نمرات
scores = [17, 14.5, 19, 12, 16.5]

total = sum(scores)          # جمع همه اعضا
count = len(scores)          # تعداد اعضا
average = total / count      # تقسیم برای میانگین

print("جمع نمرات:", total)
print("تعداد درس‌ها:", count)
print("میانگین:", average)
print("بیشترین نمره:", max(scores))
print("کمترین نمره:", min(scores))
```

### مثال ب: شمارش تکرار کلمات با دیکشنری

</div>

```python
# مثال ۸: شمارش تعداد تکرار هر میوه در یک سبد
basket = ["سیب", "موز", "سیب", "کیوی", "موز", "سیب"]

counts = {}   # دیکشنری خالی
for fruit in basket:
    # get مقدار فعلی را می‌دهد و اگر نبود صفر برمی‌گرداند
    counts[fruit] = counts.get(fruit, 0) + 1

print(counts)
```

### مثال ج: پیدا کردن دانشجویان مشترک بین دو کلاس با مجموعه‌ها

</div>

```python
# مثال ۹: اشتراک و اجتماع دو کلاس
class_a = {"علی", "سارا", "رضا", "مریم"}
class_b = {"سارا", "امید", "مریم", "نیما"}

print("دانشجویان مشترک:", class_a & class_b)
print("همه دانشجویان:", class_a | class_b)
print("فقط در کلاس A:", class_a - class_b)
print("تعداد کل دانشجویان یکتا:", len(class_a | class_b))
```

### مثال د: ترکیب رشته، تاپل و عملگرها

</div>

```python
# مثال ۱۰: مختصات دو نقطه و فاصله افقی بین آن‌ها
point_1 = (2, 3)
point_2 = (8, 11)

x1, y1 = point_1
x2, y2 = point_2

# محاسبه فاصله دو نقطه با عملگر توان (توان ۰.۵ همان جذر است)
distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

print("فاصله بین دو نقطه:", distance)

# چاپ یک خط جداکننده با تکرار رشته
print("=" * 25)
print("نقطه اول:", point_1, "| نقطه دوم:", point_2)
```
