## ۱. دریافت ورودی از کاربر (input)
* تا اینجا مقادیر را خودمان داخل کد نوشته‌ایم. با تابع `input()` می‌توانیم از کاربر اطلاعات بگیریم.
* متنی که داخل پرانتز می‌نویسیم به کاربر نمایش داده می‌شود (راهنما).
* **نکته بسیار مهم:** `input()` همیشه یک **رشته (str)** برمی‌گرداند، حتی اگر کاربر عدد وارد کند.
* در Google Colab بعد از اجرای سلول، کادری برای تایپ ظاهر می‌شود؛ مقدار را بنویسید و Enter بزنید.

</div>

```python
# دریافت نام از کاربر
name = input("نام شما چیست؟ ")
print("سلام", name)

# دریافت یک عدد (ولی به صورت متن!)
age = input("سن شما چند سال است؟ ")
print("مقدار وارد شده:", age)
print("نوع داده:", type(age))
```

## ۲. تبدیل نوع داده (Type Conversion)
* چون `input()` رشته برمی‌گرداند، برای محاسبه باید آن را به عدد تبدیل کنیم.
* توابع تبدیل نوع:
  * `int()` تبدیل به عدد صحیح
  * `float()` تبدیل به عدد اعشاری
  * `str()` تبدیل به رشته
  * `bool()` تبدیل به بولی
* اگر مقدار قابل تبدیل نباشد (مثلا `int("سلام")`) پایتون خطای `ValueError` می‌دهد؛ در بخش بعد یاد می‌گیریم چطور با این خطا برخورد کنیم.
* نکته: `"5" + "3"` برابر `"53"` است (چسباندن متن‌ها)، اما `5 + 3` برابر `8` است.

</div>

```python
# تفاوت متن و عدد
print("5" + "3")        # چسباندن دو رشته
print(5 + 3)            # جمع دو عدد

# تبدیل نوع
text_number = "25"
real_number = int(text_number)
print(real_number + 5)

print(float("3.5") * 2)
print(str(100) + " تومان")
print(int(7.9))         # اعشار حذف می‌شود (گرد نمی‌شود)
print(bool(0), bool(5), bool(""), bool("سلام"))

# استفاده همزمان از input و تبدیل نوع
year_of_birth = int(input("سال تولد شما (میلادی) چیست؟ "))
print("سن تقریبی شما:", 2026 - year_of_birth)
```

## ۳. چاپ حرفه‌ای با f-string
* برای ساختن متن‌هایی که شامل مقدار متغیرها هستند، بهترین و خواناترین روش f-string است.
* قبل از کوتیشن حرف `f` می‌گذاریم و هر متغیر یا عبارت را داخل آکولاد `{}` می‌نویسیم.
* می‌توانیم قالب نمایش اعداد را هم تعیین کنیم: `{x:.2f}` یعنی عدد با ۲ رقم اعشار، و `{x:,}` یعنی جداکننده هزارگان.

</div>

```python
name = "سارا"
score = 18.456
price = 1250000

# روش قدیمی: چسباندن و کاما
print("نام:", name, "| نمره:", score)

# روش f-string
print(f"نام: {name} | نمره: {score}")

# قالب‌بندی اعداد
print(f"نمره با دو رقم اعشار: {score:.2f}")
print(f"قیمت: {price:,} تومان")

# داخل آکولاد می‌توانیم محاسبه هم انجام دهیم
a = 4
b = 6
print(f"حاصل ضرب {a} و {b} برابر است با {a * b}")
```

## ۴. مدیریت خطا (try / except)
* وقتی پایتون به کدی برسد که نتواند اجرا کند، خطا می‌دهد و برنامه متوقف می‌شود. مثلا وقتی کاربر به جای عدد، متن وارد کند.
* با ساختار `try` و `except` می‌توانیم خطا را مدیریت کنیم تا برنامه نیفتد:
  * داخل `try` کدی را می‌نویسیم که ممکن است خطا بدهد.
  * داخل `except` مشخص می‌کنیم اگر خطا رخ داد چه کاری انجام شود.
* چند خطای رایج: `ValueError` (مقدار نامناسب)، `ZeroDivisionError` (تقسیم بر صفر)، `IndexError` (ایندکس خارج از لیست) و `KeyError` (کلید ناموجود در دیکشنری).
* ترکیب این ساختار با حلقه `while` برای گرفتن ورودی معتبر بسیار کاربرد دارد.

</div>

```python
# ۱. مدیریت خطای تبدیل نوع
try:
    number = int("abc")
    print(number)
except ValueError:
    print("خطا: این مقدار عدد نیست.")

# ۲. مدیریت تقسیم بر صفر
try:
    print(10 / 0)
except ZeroDivisionError:
    print("خطا: تقسیم بر صفر ممکن نیست.")

print("-----------------")

# ۳. گرفتن ورودی معتبر: تا زمانی که عدد درست وارد نشود، دوباره می‌پرسد
while True:
    text = input("یک عدد صحیح وارد کنید: ")
    try:
        value = int(text)
        break                 # ورودی درست بود؛ از حلقه خارج می‌شویم
    except ValueError:
        print("ورودی نامعتبر است. دوباره تلاش کنید.")

print(f"عدد وارد شده: {value}")
```

## ۵. ماژول‌ها و دستور import
* پایتون کلی کد آماده در قالب «ماژول» (فایل‌های کد) و «کتابخانه» (مجموعه‌ای از ماژول‌ها) دارد که با دستور `import` به برنامه اضافه می‌شوند.
* چند شکل نوشتن:
  * `import math` کل ماژول را می‌آورد و با `math.sqrt(...)` استفاده می‌کنیم.
  * `from math import sqrt` فقط همان تابع را می‌آورد و مستقیم `sqrt(...)` می‌نویسیم.
  * `import numpy as np` ماژول را با یک نام کوتاه‌تر (مستعار) می‌آورد.
* ماژول `math` توابع ریاضی دارد (جذر، سینوس، لگاریتم و ثابت `pi`) و ماژول `random` اعداد تصادفی تولید می‌کند.

</div>

```python
import math
import random

# ماژول math
print("جذر ۱۶:", math.sqrt(16))
print("عدد پی:", math.pi)
print("گرد به بالا:", math.ceil(4.2))
print("گرد به پایین:", math.floor(4.8))
print("لگاریتم طبیعی ۱۰:", round(math.log(10), 3))

# شکل دیگر import
from math import factorial
print("فاکتوریل ۵:", factorial(5))

print("-----------------")

# ماژول random
random.seed(42)                       # برای اینکه هر بار نتیجه یکسان باشد
print("عدد تصادفی بین ۱ تا ۱۰:", random.randint(1, 10))
print("عدد اعشاری بین ۰ و ۱:", random.random())

fruits = ["سیب", "موز", "کیوی", "انار"]
print("انتخاب تصادفی:", random.choice(fruits))
random.shuffle(fruits)                # بر زدن لیست
print("لیست بر زده شده:", fruits)
```

## ۶. آشنایی اولیه با NumPy (مخصوص هوش مصنوعی)
* NumPy مهم‌ترین کتابخانه محاسبات عددی در پایتون است و پایه تقریباً تمام کتابخانه‌های هوش مصنوعی (مثل pandas، scikit-learn و PyTorch) محسوب می‌شود.
* در Google Colab از قبل نصب شده است و فقط باید آن را import کنیم (به صورت قراردادی با نام مستعار `np`).
* ساختار اصلی آن **آرایه (array)** است؛ شبیه لیست، اما همه اعضا از یک نوع‌اند و می‌توان عملیات را **یکجا روی همه اعضا** انجام داد (بدون حلقه).
* ویژگی‌ها و توابع پرکاربرد: `.shape` (ابعاد)، `.mean()` (میانگین)، `.max()`، `.min()`، `.sum()` و `np.zeros()` برای ساخت آرایه صفر.
* آرایه دوبعدی همان ماتریس است که داده‌های جدولی در هوش مصنوعی با آن نمایش داده می‌شوند.

</div>

```python
import numpy as np

# ساخت آرایه از لیست
scores = np.array([12, 15, 18, 20, 9])
print(scores)
print("نوع:", type(scores))

# عملیات یکجا روی همه اعضا (بدون حلقه)
print("ضرب در ۲:", scores * 2)
print("اضافه کردن ۱:", scores + 1)

# توابع آماری
print("میانگین:", scores.mean())
print("بیشترین:", scores.max())
print("جمع:", scores.sum())

print("-----------------")

# آرایه دوبعدی (ماتریس)
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matrix)
print("ابعاد (سطر، ستون):", matrix.shape)
print("عضو سطر دوم ستون سوم:", matrix[1, 2])

# ساخت آرایه آماده
print(np.zeros(4))
print(np.arange(0, 10, 2))      # شبیه range
```

## ۷. مثال جامع
* در این سناریو برنامه‌ای می‌نویسیم که نام و نمرات دانشجو را از کاربر می‌گیرد و میانگین را محاسبه و گزارش می‌دهد.
* این مثال همه مفاهیم این جزوه را ترکیب می‌کند: `input` برای دریافت اطلاعات، تبدیل نوع، `try/except` برای ورودی نامعتبر، تابع برای جلوگیری از تکرار کد، f-string برای چاپ زیبا و NumPy برای محاسبه آماری.

</div>

```python
import numpy as np

def get_number(message):
    """تا زمانی که کاربر یک عدد معتبر وارد نکند، دوباره می‌پرسد و عدد را برمی‌گرداند."""
    while True:
        text = input(message)
        try:
            return float(text)
        except ValueError:
            print("خطا: لطفا فقط عدد وارد کنید.")

def get_status(avg):
    """بر اساس میانگین، وضعیت را برمی‌گرداند."""
    if avg >= 17:
        return "عالی"
    elif avg >= 12:
        return "خوب"
    elif avg >= 10:
        return "قبول"
    else:
        return "مردود"

# ۱. دریافت اطلاعات از کاربر
student_name = input("نام دانشجو: ")
course_count = int(get_number("تعداد درس‌ها: "))

scores = []
for i in range(1, course_count + 1):
    scores.append(get_number(f"نمره درس {i}: "))

# ۲. محاسبه با NumPy
scores_array = np.array(scores)
average = scores_array.mean()

# ۳. چاپ گزارش
print("-" * 30)
print(f"گزارش نمرات {student_name}")
print(f"تعداد درس‌ها: {course_count}")
print(f"بیشترین نمره: {scores_array.max()}")
print(f"کمترین نمره: {scores_array.min()}")
print(f"میانگین: {average:.2f}")
print(f"وضعیت: {get_status(average)}")
```

## ۸. مثال‌های بیشتر
### مثال الف: ماشین‌حساب BMI (شاخص توده بدنی)

</div>

```python
weight = float(input("وزن (کیلوگرم): "))
height = float(input("قد (متر): "))

bmi = weight / (height ** 2)
print(f"BMI شما برابر است با {bmi:.1f}")

if bmi < 18.5:
    print("وضعیت: کم‌وزن")
elif bmi < 25:
    print("وضعیت: طبیعی")
else:
    print("وضعیت: اضافه‌وزن")
```

### مثال ب: تبدیل دما با ورودی کاربر و مدیریت خطا

</div>

```python
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

try:
    temp = float(input("دما به سانتی‌گراد: "))
    print(f"{temp} سانتی‌گراد برابر است با {celsius_to_fahrenheit(temp):.1f} فارنهایت")
except ValueError:
    print("خطا: لطفا یک عدد وارد کنید.")
```

### مثال ج: بازی حدس عدد (random، while، input و try/except)

</div>

```python
import random

secret = random.randint(1, 20)
max_tries = 5
tries = 0

print("من یک عدد بین ۱ تا ۲۰ انتخاب کرده‌ام. ۵ بار فرصت دارید حدس بزنید!")

while tries < max_tries:
    try:
        guess = int(input(f"حدس شماره {tries + 1}: "))
    except ValueError:
        print("لطفا یک عدد صحیح وارد کنید.")
        continue                  # این حدس حساب نمی‌شود

    tries += 1

    if guess == secret:
        print(f"آفرین! عدد درست {secret} بود و در {tries} حدس پیدا کردید.")
        break
    elif guess < secret:
        print("عدد بزرگتری را امتحان کنید.")
    else:
        print("عدد کوچکتری را امتحان کنید.")
else:
    # این بخش وقتی اجرا می‌شود که حلقه بدون break تمام شده باشد
    print(f"فرصت‌ها تمام شد. عدد درست {secret} بود.")
```

### مثال د: نرمال‌سازی داده‌ها با NumPy
* در هوش مصنوعی قبل از آموزش مدل، اغلب داده‌ها را به بازه ۰ تا ۱ می‌بریم (نرمال‌سازی Min-Max). فرمول آن: `(x - min) / (max - min)`

</div>

```python
import numpy as np

ages = np.array([18, 25, 32, 47, 60])

normalized = (ages - ages.min()) / (ages.max() - ages.min())

print("داده اصلی:", ages)
print("داده نرمال‌شده:", normalized)
print(f"میانگین داده اصلی: {ages.mean():.1f}")
```

### مثال ه: محاسبه فاصله دو نقطه با ماژول math

</div>

```python
import math

x1 = float(input("x نقطه اول: "))
y1 = float(input("y نقطه اول: "))
x2 = float(input("x نقطه دوم: "))
y2 = float(input("y نقطه دوم: "))

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print(f"فاصله دو نقطه: {distance:.2f}")
```
