## ۱. عملگرهای مقایسه‌ای و منطقی
* برای تصمیم‌گیری در برنامه، اول باید بتوانیم مقادیر را با هم مقایسه کنیم. نتیجه هر مقایسه یک مقدار بولی (`True` یا `False`) است.
* **عملگرهای مقایسه‌ای:**

| عملگر | معنی | مثال | نتیجه |
| :---: | :--- | :---: | :---: |
| `==` | مساوی است با | `5 == 5` | `True` |
| `!=` | مساوی نیست با | `5 != 3` | `True` |
| `>` | بزرگتر | `3 > 5` | `False` |
| `<` | کوچکتر | `3 < 5` | `True` |
| `>=` | بزرگتر یا مساوی | `5 >= 5` | `True` |
| `<=` | کوچکتر یا مساوی | `6 <= 5` | `False` |

* **هشدار:** علامت `=` برای قرار دادن مقدار در متغیر است، اما `==` برای مقایسه به کار می‌رود. این دو را اشتباه نگیرید.
* **عملگرهای منطقی:** برای ترکیب چند شرط:
  * `and` (و): وقتی درست است که **هر دو** شرط درست باشند.
  * `or` (یا): وقتی درست است که **حداقل یکی** از شرط‌ها درست باشد.
  * `not` (نقیض): نتیجه را برعکس می‌کند.
* **عملگرهای عضویت:** `in` و `not in` بررسی می‌کنند که یک مقدار داخل لیست، رشته، مجموعه و... هست یا نه.
* **یادآوری:** عملگر `%` باقیمانده تقسیم است؛ مثلا `10 % 3` برابر `1` می‌شود.

</div>

```python
x = 8

# عملگرهای مقایسه‌ای
print(x == 8)
print(x != 8)
print(x > 10)
print(x <= 8)

# عملگرهای منطقی
print(x > 5 and x < 10)    # هر دو درست‌اند
print(x < 5 or x == 8)     # دومی درست است
print(not x > 5)           # برعکس True می‌شود

# عملگر عضویت
colors = ["قرمز", "آبی", "سبز"]
print("آبی" in colors)
print("زرد" not in colors)

# باقیمانده تقسیم و تشخیص زوج یا فرد
print(10 % 3)
print(x % 2 == 0)
```

## ۲. دستورات شرطی (if, elif, else)
* برنامه‌ها همیشه یک مسیر ثابت را طی نمی‌کنند. گاهی نیاز داریم بر اساس یک شرایط خاص، تصمیم‌گیری کنیم.
* با استفاده از کلمه `if` (اگر) یک شرط را بررسی می‌کنیم. اگر آن شرط درست باشد، کدهای زیر آن اجرا می‌شوند.
* کلمه `elif` (در غیر این صورت اگر) برای بررسی شرط‌های دوم و سوم استفاده می‌شود و زمانی چک می‌شود که شرط قبلی غلط بوده باشد.
* کلمه `else` (در غیر این صورت) زمانی اجرا می‌شود که هیچ‌کدام از شرط‌های بالا درست نباشند.
* فراموش نکنید که در پایتون بعد از نوشتن شرط، باید علامت دو نقطه `:` بگذارید و خط بعدی را با فاصله (تو رفتگی) شروع کنید تا برنامه بداند کدام کدها متعلق به آن شرط هستند.
* در شرط‌ها می‌توانیم از عملگرهای منطقی (`and`, `or`, `not`) برای ترکیب چند شرط استفاده کنیم.

</div>

```python
# تعریف یک متغیر برای بررسی نمره
score = 75

# بررسی شرایط مختلف نمره با استفاده از ساختار شرطی
if score >= 90:
    # اگر نمره بزرگتر یا مساوی ۹۰ باشد، این خط اجرا می‌شود
    print("وضعیت: عالی")
elif score >= 70:
    # اگر شرط اول غلط بود اما نمره بزرگتر یا مساوی ۷۰ باشد، این خط اجرا می‌شود
    print("وضعیت: خوب")
else:
    # اگر هیچ‌کدام از شرط‌های بالا درست نبود، برنامه به اینجا می‌رسد
    print("وضعیت: نیاز به تلاش بیشتر")

print("-----------------")

# ترکیب چند شرط با and
age = 25
has_license = True

if age >= 18 and has_license:
    print("می‌توانید رانندگی کنید.")
else:
    print("نمی‌توانید رانندگی کنید.")
```

## ۳. حلقه‌های تکرار (For Loop)
* در برنامه‌نویسی خیلی اوقات نیاز داریم یک کار را چندین بار تکرار کنیم یا روی مجموعه‌ای از داده‌ها (مثل یک لیست از اسامی) حرکت کنیم.
* حلقه `for` دقیقا برای همین کار ساخته شده است. این حلقه به تعداد اعضای یک مجموعه، کدهای داخل خود را تکرار می‌کند.
* معمولا از دستور `range()` در کنار حلقه `for` استفاده می‌کنیم تا یک بازه عددی تولید کنیم. سه شکل دارد:
  * `range(5)` یعنی اعداد ۰ تا ۴
  * `range(1, 6)` یعنی از ۱ شروع کن تا قبل از ۶ (یعنی ۱ تا ۵)؛ عدد پایان شامل نمی‌شود
  * `range(0, 10, 2)` یعنی از ۰ تا قبل از ۱۰ با گام ۲ (یعنی ۰، ۲، ۴، ۶، ۸)
* اگر هم شماره و هم مقدار عضو را بخواهیم از `enumerate()` استفاده می‌کنیم.

</div>

```python
# ۱. پیمایش یک لیست با حلقه for
names = ["علی", "سارا", "رضا"]

# کلمه name در اینجا به ترتیب مقدار علی، سارا و رضا را به خود می‌گیرد
for name in names:
    print("سلام", name)

print("-----------------")

# ۲. استفاده از حلقه و دستور range برای تکرار به تعداد مشخص
# این کد اعداد ۰ تا ۲ را چاپ می‌کند (عدد ۳ شامل نمی‌شود)
for number in range(3):
    print("تکرار شماره:", number)

print("-----------------")

# ۳. range با شروع و پایان و گام
for number in range(1, 6):
    print(number, end=" ")      # end=" " باعث می‌شود همه در یک خط چاپ شوند
print()

for number in range(0, 10, 2):
    print(number, end=" ")
print()

print("-----------------")

# ۴. enumerate: دریافت شماره و مقدار با هم
for index, name in enumerate(names):
    print("شماره", index, ":", name)

# ۵. پیمایش یک رشته
for letter in "Python":
    print(letter, end="-")
print()
```

## ۴. حلقه‌های شرطی (While Loop)
* حلقه `while` (تا زمانی که) نوع دیگری از حلقه‌ها است که بر اساس یک شرط کار می‌کند، نه یک تعداد مشخص.
* این حلقه تا زمانی که شرط جلوی آن درست (True) باشد، کدهای داخل خود را مدام تکرار می‌کند.
* نکته بسیار مهم در حلقه `while` این است که باید در داخل بدنه حلقه کاری کنیم که شرط در نهایت غلط شود (مثلا یک متغیر را تغییر دهیم)، وگرنه حلقه تا بی‌نهایت ادامه پیدا می‌کند و برنامه هنگ می‌کند.
* تفاوت `for` و `while`: وقتی تعداد تکرار را می‌دانیم یا روی یک مجموعه حرکت می‌کنیم از `for` و وقتی تعداد تکرار نامشخص است از `while` استفاده می‌کنیم.

</div>

```python
# تعیین مقدار اولیه شمارنده
counter = 1

# حلقه تا زمانی که شمارنده کوچکتر یا مساوی ۳ باشد ادامه دارد
while counter <= 3:
    # چاپ مقدار فعلی شمارنده
    print("مقدار شمارنده در حال حاضر:", counter)

    # اضافه کردن یک واحد به شمارنده برای جلوگیری از حلقه بی‌نهایت
    counter += 1

print("حلقه با موفقیت به پایان رسید.")

print("-----------------")

# مثال دیگر: جمع اعداد ۱ تا ۵ با while
total = 0
n = 1
while n <= 5:
    total += n
    n += 1
print("مجموع اعداد ۱ تا ۵:", total)
```

## ۵. کنترل جریان حلقه‌ها (Break و Continue)
* گاهی اوقات می‌خواهیم رفتار طبیعی یک حلقه را تغییر دهیم. پایتون دو دستور بسیار مفید برای این کار در اختیار ما گذاشته است.
* دستور `break` (شکستن) کل حلقه را بلافاصله متوقف می‌کند و برنامه کاملا از حلقه خارج می‌شود. این دستور برای زمان‌هایی کاربرد دارد که به هدفمان رسیده‌ایم و دیگر نیازی به ادامه حلقه نداریم.
* دستور `continue` (ادامه دادن) فقط تکرار فعلی (همین دور از حلقه) را متوقف می‌کند و مستقیم به سراغ دور بعدی حلقه می‌رود. این دستور برای نادیده گرفتن برخی مقادیر خاص استفاده می‌شود.

</div>

```python
# مثال برای دستور break
for num in range(1, 6):
    if num == 4:
        # وقتی عدد به ۴ برسد، حلقه کاملا متوقف می‌شود و به ۵ نمی‌رسد
        break
    print("چاپ با دستور break:", num)

print("-----------------")

# مثال برای دستور continue
for num in range(1, 6):
    if num == 3:
        # وقتی عدد به ۳ برسد، این دور رها شده و عدد ۳ چاپ نمی‌شود، اما حلقه برای ۴ و ۵ ادامه می‌یابد
        continue
    print("چاپ با دستور continue:", num)
```

## ۶. حلقه‌های تو در تو
* می‌توانیم یک حلقه را داخل حلقه دیگر قرار دهیم. به ازای هر دور از حلقه بیرونی، حلقه داخلی یک بار کامل اجرا می‌شود.
* تو رفتگی حلقه داخلی باید یک سطح (۴ فاصله) بیشتر از حلقه بیرونی باشد.
* این روش برای کار با جدول‌ها و ماتریس‌ها (که در هوش مصنوعی زیاد استفاده می‌شوند) کاربرد دارد.

</div>

```python
# جدول ضرب ۱ تا ۳
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end="\t")    # \t یعنی یک فاصله تب
    print()                       # رفتن به خط بعد بعد از پایان هر سطر
```

## ۷. ساخت سریع لیست‌ها (List Comprehensions)
* پایتون یک روش بسیار کوتاه، زیبا و حرفه‌ای برای ساخت لیست‌ها دارد که به آن List Comprehension می‌گویند.
* این روش به ما اجازه می‌دهد کدی که در حالت عادی نیاز به چند خط حلقه و دستور شرطی دارد را، تنها در یک خط بنویسیم.
* شکل کلی آن: `[عبارت  for متغیر in مجموعه  if شرط]` که بخش `if شرط` اختیاری است.
* این کار باعث تمیزی و سرعت بیشتر کد می‌شود، اما به عنوان یک فرد مبتدی، بهتر است ابتدا با روش عادی راحت شوید و سپس از این روش برای کدهای ساده استفاده کنید.

</div>

```python
# حالت عادی: ساخت لیستی از اعداد زوج بین ۱ تا ۱۰
even_numbers = []
for x in range(1, 11):
    if x % 2 == 0:
        even_numbers.append(x)
print("ساخت لیست به روش عادی:", even_numbers)

# حالت حرفه‌ای: ساخت همان لیست بالا فقط در یک خط خلاصه
# ترجمه کد: به ازای هر x از ۱ تا ۱۰، اگر x بر ۲ بخش‌پذیر است، x را در لیست بگذار
fast_even_numbers = [x for x in range(1, 11) if x % 2 == 0]
print("ساخت لیست به روش سریع:", fast_even_numbers)

# بدون شرط: لیست مربع اعداد ۱ تا ۵
squares = [x ** 2 for x in range(1, 6)]
print("مربع‌ها:", squares)
```

## ۸. مثال جامع
* در این سناریو، ما سیستمی برای بررسی نمرات پایان‌ترم چند دانش‌آموز می‌نویسیم.
* در این کد از حلقه `for` برای بررسی تک‌تک نمرات، از شرط‌های `if` و `else` برای تعیین وضعیت قبولی یا ردی، و از دستور `continue` برای نادیده گرفتن نمرات نامعتبر (نمرات منفی که اشتباه ثبت شده‌اند) استفاده کرده‌ایم.
* این مثال به خوبی نشان می‌دهد که چگونه تمام ابزارهای کنترل جریان برنامه با هم ترکیب می‌شوند تا یک مسئله واقعی را حل کنند.

</div>

```python
# لیستی از نمرات دانش‌آموزان (عدد منفی نشانه خطای ثبت است)
student_scores = [85, 45, -10, 92, 60]

# متغیری برای نگهداری تعداد افراد قبول شده
passed_count = 0

# شروع حلقه برای بررسی تک‌تک نمرات لیست
for score in student_scores:

    # بررسی اعتبار نمره: اگر نمره منفی بود، یعنی اشتباه وارد شده است
    if score < 0:
        print("خطا: نمره نامعتبر پیدا شد. از بررسی این مورد عبور می‌کنیم.")
        # دستور continue باعث می‌شود کدهای پایین اجرا نشوند و حلقه به سراغ نمره بعدی برود
        continue

    # بررسی شرط قبولی (نمره بالای ۵۰)
    if score >= 50:
        print("نمره", score, "-> قبول")
        # اضافه کردن یک واحد به متغیر افراد قبول شده
        passed_count += 1
    else:
        print("نمره", score, "-> رد شده")

# چاپ نتیجه نهایی پس از پایان کار حلقه
print("تعداد کل دانش‌آموزان قبول شده در این کلاس:", passed_count)
```

## ۹. مثال‌های بیشتر
### مثال الف: پیدا کردن بزرگترین عدد لیست (بدون استفاده از max)

</div>

```python
numbers = [12, 45, 7, 89, 23, 56]

# فرض می‌کنیم عضو اول بزرگترین است
biggest = numbers[0]

for n in numbers:
    # هر عددی که از بزرگترین فعلی بزرگتر بود جایگزین می‌شود
    if n > biggest:
        biggest = n

print("بزرگترین عدد:", biggest)
```

### مثال ب: بررسی عدد اول بودن (استفاده از for، if، break و %)

</div>

```python
number = 29
is_prime = True

# اعداد کمتر از ۲ اول نیستند
if number < 2:
    is_prime = False

# بررسی بخش‌پذیری بر اعداد ۲ تا قبل از خود عدد
for d in range(2, number):
    if number % d == 0:
        # یک مقسوم‌علیه پیدا شد؛ پس عدد اول نیست و ادامه دادن لازم نیست
        is_prime = False
        break

if is_prime:
    print(number, "عدد اول است.")
else:
    print(number, "عدد اول نیست.")
```

### مثال ج: شمارش زوج و فرد با حلقه while

</div>

```python
n = 1
even_count = 0
odd_count = 0

while n <= 10:
    if n % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
    n += 1

print("تعداد زوج‌ها:", even_count)
print("تعداد فردها:", odd_count)
```

### مثال د: فیلتر کردن لیست با شرط ترکیبی و List Comprehension

</div>

```python
ages = [12, 25, 17, 40, 18, 9, 33]

# افراد بزرگسال (۱۸ تا ۳۵ سال) را جدا می‌کنیم
adults = [a for a in ages if a >= 18 and a <= 35]
print("بزرگسالان ۱۸ تا ۳۵ سال:", adults)

# تعداد افراد زیر ۱۸ سال
minors_count = len([a for a in ages if a < 18])
print("تعداد زیر ۱۸ سال:", minors_count)
```
